```python
score = input('Enter your score: ')
score = int(score)
if score >= 0 and score <= 100:
    if score > 90 and score < 100:
        print('A')
    elif score > 80 and score < 91:
        print('B')
    elif score > 70 and score < 81:
        print('C')
    elif score > 60 and score < 71:
        print('D')
    elif score > 50 and score < 61:
        print('E')
    else:
        print('F')
```
